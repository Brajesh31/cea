# Sigmoid and majority-class baseline
This optional notebook revisits two concepts from the Streamlit demos. All values are synthetic.


```python
import numpy as np
import pandas as pd
```

## Sigmoid activation
The function maps real-valued inputs into the open interval (0, 1). This cell is a mathematical illustration, not a complete neural-network training pipeline.


```python
def sigmoid(z):
    z = np.asarray(z, dtype=float)
    return 1.0 / (1.0 + np.exp(-z))

values = np.array([-2.0, 0.0, 2.0])
sigmoid(values)
```

## Majority-class baseline
A constant classifier predicts the most frequent training class. Evaluation on the same labels used to select that class is in-sample only.


```python
labels = pd.Series([2, 1, 2, 0, 2, 1, 2])
majority_class = labels.value_counts().idxmax()
predictions = np.full(len(labels), majority_class)
in_sample_accuracy = np.mean(predictions == labels.to_numpy())
majority_class, in_sample_accuracy
```
