# Sigmoid and majority-class baseline
Optional companion notebook. All values are synthetic.


```python
import numpy as np
import pandas as pd
```

## Sigmoid activation
Mathematical illustration, not a complete training pipeline.


```python
def sigmoid(z):
    z = np.asarray(z, dtype=float)
    return 1.0 / (1.0 + np.exp(-z))

values = np.array([-2.0, 0.0, 2.0])
sigmoid(values)
```

## Majority-class baseline
Accuracy on the labels used to select the majority class is in-sample only.


```python
labels = pd.Series([2, 1, 2, 0, 2, 1, 2])
majority_class = labels.value_counts().idxmax()
predictions = np.full(len(labels), majority_class)
in_sample_accuracy = np.mean(predictions == labels.to_numpy())
majority_class, in_sample_accuracy
```
